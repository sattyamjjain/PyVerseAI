```python
from IPython.display import display, Markdown
from src.load_data import load_requirements, load_products
```

### requirements

```python
requirements = load_requirements()
products = load_products()
```

```python
n = 5
for _, row in requirements.head(n).iterrows():
    display(Markdown(f"**{row['requirement'].strip()}**\n\n{row['requirement_detail']}\n\n"))
```

### requirements and selected products

```python
merged = requirements.merge(products, on='product_id', how='left')
n = 5
for _, row in merged.head(n).iterrows():
    display(Markdown(f"**{row['requirement'].strip()}**\n\n{row['requirement_detail']}\n\n_Matching Product:_\n\n{row['name']}\n\n{row['description']}\n\n"))
```
